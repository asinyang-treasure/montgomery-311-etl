```python

import requests
url = "https://data.montgomerycountymd.gov/resource/xtyh-brr2.json"


response = requests.get(url, {"$limit":5, "$offset":0})
data = response.json() 
print(data[0])
print("\n")

for d in data:  
    print("Serial Number: ", d['sr_num'])
    print("Date Created: ", d['created'])
    print("Status: ", d['sr_stat_id'])
    print("Location: ", d.get('x_city', 'No value found') + ", " + d.get('x_state', 'No value found'))
    print('\n\n')

```
